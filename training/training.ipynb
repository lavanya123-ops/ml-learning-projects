# Encoding
    -Encoding is the Process of Converting String value to Numerical Values.
    -String Values are also called Categorical Values.
    -Categorical Values are devided into 2 types ->Nominal value and Ordinal Value.
    -Nominal values->cannot give any order or rank.
    -Ordinal values->can give order or rank.
    -For Ordinal values,we can go with ->Label Encoder or Mapping ->using Scikit-learn library
    -For Nominal values,we can go with ->One Hot Encoding or Dummy Encoding ->Scikit-learn library(one-hot) and pandas(dummy)

```python
def predict_stroke(gender='Male',age=56,hypertension='Yes',heart_disease='Yes',ever_married='Yes',work_type='Private',avg_glucose_level=123.45,bmi=78.4,smoking_status='smokes',resident_type='Urban'):
  lst=[]
  if gender=='Male':
    lst=lst+[0]
  elif gender=='Female':
    lst=lst+[1]
  elif gender=='Other':
    lst=lst+[2]
  lst=lst+[age]

  if hypertension=='Yes':
    lst=lst+[1]
  elif hypertension=='No':
    lst=lst+[0]
  if heart_disease=='Yes':
    lst=lst+[1]
  elif heart_disease=='No':
    lst=lst+[0]
  if ever_married=='Yes':
    lst=lst+[1]
  elif ever_married=='No':
    lst=lst+[0]    
      
  work_type=lb_worktype.transform([work_type])
  lst=lst+list(work_type)
  lst=lst+[avg_glucose_level,bmi]
  smoking_status=lb_smoking.transform([smoking_status])
  lst=lst+list(smoking_status)
  if resident_type=='Rural':
      lst=lst+[1,0]
  elif resident_type=='Urban':
      lst=lst+[0,1] 
  result=rf_model.predict([lst])
  if result==[0]:
      return "Person is not having the stroke"
  elif result==[1]:
      return "Person suffering from the stroke"
  
```
